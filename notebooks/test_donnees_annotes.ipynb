```python
import pandas as pd

df = pd.read_parquet(
    "../data/vector_store/event_documents.parquet"
)

columns = [
    "uid",
    "title",
    "dateRange",
    "firstTiming",
    "nextTiming",
    "conditions",
    "registration",
    "billetterie",
    "links",
]

available_columns = [
    col for col in columns
    if col in df.columns
]

events = (
    df[available_columns]
    .drop_duplicates(subset="uid")
)

print(events.head(30).to_string(index=False))
```

```python
booking_events = events[
    events["registration"].notna()
    & events["billetterie"].notna()
    & events["conditions"].notna()
    & events["nextTiming"].notna()
]

print(
    booking_events[
        [
            "title",
            "dateRange",
            "nextTiming",
            "conditions",
            "registration",
            "billetterie",
            "links",
        ]
    ]
    .head(20)
    .to_string(index=False)
)
```
